```python
import numpy as np
import matplotlib.pyplot as plt

# 假设 shape = (25472, 14), 即 (n_times, n_channels)
eeg = one_trial["eeg_stim"]
sfreq = 128.0  # 按你的真实采样率改

n_times, n_ch = eeg.shape
t = np.arange(n_times) / sfreq

# 画前10秒，避免太密
sec = 100
n_show = int(sec * sfreq)
x = t[:n_show]
y = eeg[:n_show, :]   # (n_show, 14)

# 多通道“堆叠”静态图（每个通道加一个垂直偏移）
offset = 3 * np.std(y)   # 通道间距
plt.figure(figsize=(14, 8))

for ch in range(n_ch):
    plt.plot(x, y[:, ch] + ch * offset, lw=0.8, color="black")
    plt.text(x[-1] + 0.05, ch * offset, f"Ch{ch+1}", va="center", fontsize=8)

plt.xlabel("Time (s)")
plt.ylabel("Channels (offset)")
plt.title('Static EEG plot: one_trial["eeg_stim"] (first 10s)')
plt.grid(alpha=0.2)
plt.tight_layout()
plt.show()
```
